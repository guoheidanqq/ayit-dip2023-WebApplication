## 3. 服务启动

在 AI Studio Terminal 中进入项目目录：

```bash
cd ~/ayit-dip2023-WebApplication/
```

执行启动命令：

```bash
uvicorn backend.main:app --reload --host 0.0.0.0 --port 8000
```

服务启动成功，终端输出：

```text
INFO: Will watch for changes in these directories:
['/home/aistudio/ayit-dip2023-WebApplication']

INFO: Uvicorn running on http://0.0.0.0:8000
(Press CTRL+C to quit)

INFO: Started reloader process [212100] using StatReload
INFO: Started server process [212102]
INFO: Waiting for application startup.
INFO: Application startup complete.
```

### 启动结果

- Uvicorn 成功启动。
- FastAPI 应用初始化完成。
- 服务监听 8000 端口。
- 开发模式自动重载功能正常。

## 4. FastAPI 接口测试

```python
@app.get("/api/hello")
def hello():
    return {
        "message": "Hello, DIPWeb!",
        "status": "success"
    }
```

测试接口：
